## جای این صفحه در درس

بندهای ۱۴.۳.۴ و ۱۴.۳.۵ از {cite}`hastie2009esl` (صص ۵۰۷–۵۰۹) پل میان ماتریس نامتشابهی و الگوریتم‌هاست. دو کار می‌کنند: روش‌های خوشه‌بندی را بر پایهٔ **آنچه هر روش دربارهٔ داده فرض می‌کند** به سه خانواده دسته می‌کنند، و بعد خانوادهٔ ترکیبیاتی را به‌صورت یک مسئلهٔ بهینه‌سازی صریح می‌نویسند، همان صورتی که هر روش صفحه‌های بعد حالت خاصی از آن است.

این بهینه‌سازی جواب دقیق دارد — شمردن همهٔ افرازها — که به‌طور بی‌فایده‌ای گران است. آنچه می‌ماند فرود است، و فرود هم قوت و هم ضعف K-means را توضیح می‌دهد.

### در پایان باید بتوانید

* بگویید یک الگوریتم داده‌شده به کدام یک از سه خانواده می‌رسد و در نتیجه چه فرضی می‌گذارد؛
* پراکندگی درون‌خوشه‌ای $W(C)$ و میان‌خوشه‌ای $B(C)$ را بنویسید و $T = W + B$ را اثبات کنید؛
* بگویید چرا کمینه‌کردن $W$ همان بیشینه‌کردن $B$ است؛
* توضیح دهید چرا هر روش عملی یک فرود حریص است که می‌تواند در کمینهٔ محلی بایستد.


## ۱. سه خانواده

| خانواده | چه فرضی می‌کند | چه چیزی برآورد می‌شود | مثال‌ها |
|---|---|---|---|
| **ترکیبیاتی** | چیزی بیش از نامتشابهی‌ها | انتساب $C(i)$ هر مشاهده | K-means، K-medoids، ادغام سلسله‌مراتبی |
| **مدل مخلوط** | داده نمونه‌ای i.i.d. از مخلوطی از چگالی‌هاست | پارامترهای چگالی‌های مؤلفه | مخلوط‌های گاوسی، خوشه‌بندی مدل‌بنیان |
| **جست‌وجوی مُد** | چگالی نامعلوم است ولی مُدهای مشخص دارد | مُدها، و سپس نقطه‌های نزدیک هر مُد | PRIM («شکار برجستگی»)، mean-shift، DBSCAN |

این تقسیم دربارهٔ فرض‌هاست، نه کیفیت. روش ترکیبیاتی هرگز نمی‌پرسد داده از کجا آمده؛ مدل مخلوط مدلی احتمالاتی می‌گوید و آن را با درست‌نمایی بیشینه برآورد می‌کند؛ جست‌وجوی مُد چگالی را ناپارامتری برآورد و همسایگی هر مُد را یک خوشه می‌نامد. خانوادهٔ اول در عمل بسیار بزرگ‌ترین است و همان است که این درس گسترش می‌دهد؛ مدل مخلوط در ادامهٔ درس جداگانه می‌آید و جست‌وجوی مُد فقط به‌عنوان ایدهٔ پشت روش‌های چگالی‌بنیان ظاهر می‌شود.


## ۲. صورت‌بندی ترکیبیاتی

هر مشاهده با عدد صحیح $i \in \{1,\dots,N\}$ و هر خوشه با $k \in \{1,\dots,K\}$ برچسب می‌خورد، با $K < N$ که از پیش داده شده است. یک خوشه‌بندی یک کدگذار است:

$$
k = C(i),
$$

نگاشتی چندبه‌یک که هر مشاهده را به یک و تنها یک خوشه می‌فرستد. **پارامترهای مسئله همان $N$ برچسب‌اند.** مدلی برای برازش و چیزی برای مشتق‌گیری نیست؛ مجموعه‌ای متناهی از انتساب‌های نامزد و یک تابع زیان هست که می‌گوید هرکدام چقدر بد است.

زیان طبیعی، مجموع نامتشابهی *درون* خوشه‌هاست:

$$
W(C) = \frac{1}{2} \sum_{k=1}^{K} \sum_{C(i)=k} \sum_{C(i')=k} d(x_i, x_{i'}),
\qquad \text{(14.28)}
$$

همان «پراکندگی نقطه‌ای درون‌خوشه‌ای»: وقتی نقطه‌هایی که با هم در یک خوشه‌اند نزدیک باشند کوچک است. شکستن پراکندگی کل به بخش درونی و بخش میان‌خوشه‌ای می‌دهد:

$$
T = \frac{1}{2}\sum_{i=1}^{N}\sum_{i'=1}^{N} d_{ii'} = W(C) + B(C),
\qquad
B(C) = \frac{1}{2} \sum_{k=1}^{K} \sum_{C(i)=k} \sum_{C(i') \ne k} d_{ii'} .
\qquad \text{(14.29)}
$$

$T$ فقط به داده بستگی دارد، پس **کمینه‌کردن $W$ دقیقاً بیشینه‌کردن $B$ است**. این دو بیان در جاهای مختلف به کار می‌آیند: «خوشه‌ها را جمع‌وجور کن» و «خوشه‌ها را از هم دور کن» یک دستورند.

سلول زیر این اتحاد را روی مجموعه‌داده‌ای شبیه‌سازی‌شده و انتسابی تصادفی بررسی می‌کند.


```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
centres = [(-3.0, 0.0), (3.0, 0.0), (0.0, 4.0), (0.0, -4.0)]   # four groups
X = np.vstack([rng.normal(c, 0.7, size=(15, 2)) for c in centres])
N = len(X)

dist = ((X[:, None, :] - X[None, :, :]) ** 2).sum(-1)   # squared Euclidean


def W_of(C):
    same = C[:, None] == C[None, :]
    return 0.5 * dist[same].sum()


def B_of(C):
    different = C[:, None] != C[None, :]
    return 0.5 * dist[different].sum()


C_random = rng.integers(0, 3, size=N)
T = 0.5 * dist.sum()
print(f"W(C)      = {W_of(C_random):9.2f}")
print(f"B(C)      = {B_of(C_random):9.2f}")
print(f"W(C) + B  = {W_of(C_random) + B_of(C_random):9.2f}")
print(f"T         = {T:9.2f}")
```

### جزئیاتی که ارزش دانستن دارد: $W$ و inertia در `sklearn`

با فاصلهٔ اقلیدسی مجذور، صورت جفتی (۱۴.۲۸) را می‌توان برحسب میانگین‌های خوشه بازنویسی کرد و همین است که K-means را ارزان می‌کند:

$$
\frac{1}{2}\sum_{i \in C_k}\sum_{i' \in C_k} \lVert x_i - x_{i'} \rVert^2
= N_k \sum_{i \in C_k} \lVert x_i - \bar{x}_k \rVert^2 .
\qquad \text{(14.31)}
$$

سمت چپ هر جفت را یک‌بار می‌شمارد؛ سمت راست مجموع مربعات خوشه را در اندازه‌اش $N_k$ ضرب می‌کند. پس $W$ همان عددی *نیست* که `KMeans.inertia_` گزارش می‌کند — آن $\sum_k \sum_{i \in C_k}\lVert x_i - \bar x_k\rVert^2$ است، بدون $N_k$ — و لازم نیست کمینه‌ساز یکسان داشته باشند. هر دو پراکندگی درون‌خوشه‌ای‌اند، هر دو با جمع‌وجورتر شدن خوشه‌ها کم می‌شوند، و کتاب از (۱۴.۳۱) بهره می‌برد چون جمعی $N^2$ تایی را به جمعی $N$ تایی تبدیل می‌کند. این ضریب را وقتی عددهای منابع مختلف را مقایسه می‌کنید در یاد نگه دارید.


```python
def means_form(C):
    total = 0.0
    for k in np.unique(C):
        members = X[C == k]
        total += len(members) * ((members - members.mean(axis=0)) ** 2).sum()
    return total


plain_ss = sum(((X[C_random == k] - X[C_random == k].mean(axis=0)) ** 2).sum()
               for k in np.unique(C_random))
print("pairwise W(C)                   :", round(W_of(C_random), 2))
print("sum_k N_k * SS_k, form (14.31)  :", round(means_form(C_random), 2))
print("sum_k SS_k, the sklearn inertia :", round(plain_ss, 2))
```

## ۳. شمردن دقیق است و بی‌فایده

تعداد راه‌های شکستن $N$ مشاهدهٔ برچسب‌دار به $K$ خوشهٔ ناتهی همان عدد استرلینگ نوع دوم است:

$$
S(N,K) = \frac{1}{K!}\sum_{k=1}^{K}(-1)^{K-k}\binom{K}{k}k^{N} .
\qquad \text{(14.30)}
$$

برای $N = 10$ و $K = 4$ این $34\,105$ است که یک لپ‌تاپ می‌تواند بشمارد. برای $N = 19$ از قبل $10^{10}$ است، و مسئله‌های خوشه‌بندی از صدها مشاهده آغاز می‌شوند. شمارش، کد شمردن و هرس‌کردنی که نمونه‌های کوچک را قابل‌حل می‌کند در [صفحهٔ خوشه‌بندی فراگیر](Exhaustive-Clustering.ipynb) کار شده است؛ اینجا فقط پیامد مهم است:


```python
from math import comb, factorial


def stirling2(n, k):
    return sum((-1) ** (k - j) * comb(k, j) * j ** n for j in range(k + 1)) // factorial(k)


for n, k in [(10, 4), (19, 4), (50, 3)]:
    print(f"S({n:>2}, {k}) = {stirling2(n, k):.3e}")
```

## ۴. آنچه می‌ماند: فرود حریص

یک روش عملی از افرازی آغازین شروع می‌کند و تا وقتی تغییری $W$ را بهتر می‌کند برچسب‌ها را عوض می‌کند. وقتی هیچ تغییر تکی آن را بهتر نکند، الگوریتم می‌ایستد و انتساب کنونی را برمی‌گرداند. این یک کمینهٔ **محلی** از تابعی است که روی مجموعه‌ای متناهی ولی نجومی تعریف شده و هیچ‌چیز در رویه نمی‌گوید کمینهٔ جهانی نزدیک است.

کد زیر دقیقاً همین کار را می‌کند، با حرکت خام «اگر $W$ کم شود، یک نقطه را به خوشهٔ دیگری ببر»، که روی ماتریس جفتی کامل ارزیابی می‌شود. برای دادهٔ واقعی بسیار کند است — از همین رو K-means آن را با حرکت دوگامی «میانگین‌ها، بعد نزدیک‌ترین میانگین» جانشین می‌کند — ولی شکل هر روش این فصل را نشان می‌دهد.


```python
def wcss(X, C):
    """Sum over clusters of the sum of squared distances to the cluster mean."""
    return sum(((X[C == k] - X[C == k].mean(axis=0)) ** 2).sum() for k in np.unique(C))


def wcss_moves(X, C, K):
    cur = wcss(X, C)
    for i in range(len(X)):
        for k in range(K):
            if k == C[i]:
                continue
            trial = C.copy()
            trial[i] = k
            if len(np.unique(trial)) < K:      # keep K non-empty clusters
                continue
            if wcss(X, trial) < cur - 1e-9:
                return i, k
    return None


def greedy(X, K, seed, max_steps=400):
    rng = np.random.default_rng(seed)
    C = rng.integers(0, K, size=len(X))
    history = [wcss(X, C)]
    for _ in range(max_steps):
        move = wcss_moves(X, C, K)
        if move is None:
            break
        C[move[0]] = move[1]
        history.append(wcss(X, C))
    return C, history


for seed in range(5):
    C, history = greedy(X, 3, seed)
    print(f"start {seed}: steps {len(history) - 1:3d}   "
          f"W from {history[0]:8.2f} to {history[-1]:8.2f}")
```

هر سطر کاهش می‌یابد و سطرها همه به یک مقدار نمی‌رسند: با چهار گروه و $K = 3$، یکی از پنج شروع روی خوشه‌بندی‌ای ایستاد که هیچ حرکت تکی بهترش نمی‌کند و با این حال از آنچه بقیه یافتند بدتر است. این یک کمینهٔ محلی از معیاری با چند حوضهٔ جذب است، و مهم‌ترین واقعیت عملی دربارهٔ خوشه‌بندی: نقطهٔ آغاز بخشی از جواب است؛ از همین رو هر پیاده‌سازی الگوریتم را از چند شروع اجرا می‌کند و بهترین را نگه می‌دارد — و از همین رو تعداد شروع‌ها بخشی از چیزی است که گزارش می‌کنید.

K-means همین طرح است با فاصلهٔ اقلیدسی مجذور و حرکتی هوشمندانه‌تر: با میانگین‌های کنونی، هر نقطه به نزدیک‌ترین میانگین می‌رود، که فقط $N$ محاسبهٔ فاصله می‌خواهد نه $N^2$. معیاری که پایین می‌آورد همان پراکندگی درون‌خوشه‌ای (۱۴.۳۱) است؛ بهای حرکت ارزان‌تر این است که تنها زیرمجموعهٔ کوچکی از بازبرچسب‌گذاری‌های ممکن را می‌بیند، پس آن هم می‌تواند زود بایستد. [خوشه‌بندی فراگیر](Exhaustive-Clustering.ipynb) این دو را روی مسئله‌ای مقایسه می‌کند که به‌اندازهٔ کافی کوچک است تا بهینهٔ جهانی‌اش حساب شود.


## خلاصه

* روش‌های خوشه‌بندی بر پایهٔ *آنچه فرض می‌کنند* به خانواده‌های ترکیبیاتی، مدل مخلوط و جست‌وجوی مُد تقسیم می‌شوند؛ روش ترکیبیاتی فقط نامتشابهی‌ها را فرض می‌کند.
* در خانوادهٔ ترکیبیاتی پارامترها خودِ انتساب‌ها هستند و زیان، پراکندگی درون‌خوشه‌ای $W(C)$ از (۱۴.۲۸) است.
* $T = W + B$ با $T$ ثابت، پس کمینه‌کردن $W$ و بیشینه‌کردن $B$ یک مسئله‌اند.
* جواب دقیق $S(N,K)$ ارزیابی می‌خواهد، که از قبل در $N = 19$ و $K = 4$ برابر $10^{10}$ است.
* روش‌های عملی فرودهای حریص‌اند که در کمینهٔ محلی می‌ایستند، و نقطهٔ شروع تعیین می‌کند کدام یکی.


## تمرین‌ها

1. $T = W + B$ را از تعریف‌ها اثبات کنید و آن را برای $K = 1$ و $K = N$ بررسی کنید.
2. (۱۴.۳۱) را برای خوشه‌ای سه‌نقطه‌ای دستی و بعد با دو خط numpy بررسی کنید.
3. نشان دهید وقتی همهٔ خوشه‌ها اندازهٔ یکسان دارند، $W$ و مجموع مربعات بی‌وزن مقدار یکسان می‌دهند. کِی دو افراز را متفاوت رتبه‌بندی می‌کنند؟
4. $S(N,2) = 2^{N-1} - 1$ را از (۱۴.۳۰) حساب کنید و با [خوشه‌بندی فراگیر](Exhaustive-Clustering.ipynb) مقایسه کنید.
5. `greedy` را چنان تغییر دهید که نخستین حرکت بهبوددهنده را بپذیرد نه بهترین را. اجراها گام کمتری برمی‌دارند؟ پایین‌تر می‌ایستند یا بالاتر؟
6. دو مجموعه‌داده مثال بدهید که برای هر افراز $W$ یکسان داشته باشند ولی جواب «کدام خوشه‌بندی درست است» در آن‌ها متفاوت باشد.
